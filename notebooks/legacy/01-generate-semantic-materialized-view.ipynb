```python
from pysus import ibge
import pandas as pd
import os

# 1. Obter dados de população para municípios usando pysus
# A função get_population do PySUS pode retornar a população por município.
# Por padrão, ela busca a população mais recente ou para um ano específico.
# Para municípios, a granularidade deve ser 'municipio'.
# Usaremos o ano do último censo (2022) como referência.

pop_files = ibge(year=2022)

# Check if pop_files is a list of file paths and load them
if isinstance(pop_files, list) and pop_files:
    try:
        # Concatenate all data files into a single DataFrame
        pop_df = pd.concat([pd.read_parquet(f) for f in pop_files])
        print(f"DataFrame loaded from {len(pop_files)} files.")
    except Exception as e:
        print(f"Error loading data from paths in pop_files: {e}")
        pop_df = pd.DataFrame() # Initialize as empty DataFrame on error
elif isinstance(pop_files, pd.DataFrame):
    # If ibge already returns a DataFrame, use it directly
    pop_df = pop_files
else:
    print(f"Unexpected type for pop_files: {type(pop_files)}. Initializing empty DataFrame.")
    pop_df = pd.DataFrame()

# Proceed only if pop_df is not empty
if not pop_df.empty:
    print("Colunas do DataFrame após get_population:", pop_df.columns.tolist())

    # 2. Limpeza e formatação dos dados (PySUS já deve retornar um DataFrame limpo)
    # Renomear as colunas para o formato esperado
    pop_df = pop_df.rename(columns={
        'MUNIC_RES': 'Codigo_Municipio',
        'municipio': 'Municipio',
        'POPULACAO': 'Populacao' # Corrected case from 'populacao' to 'POPULACAO'
    })

    print("Colunas do DataFrame após rename:", pop_df.columns.tolist())

    # Garantir que a coluna 'Codigo_Municipio' é numérica e de 6 dígitos para o merge futuro
    # O SINAN usa códigos de 6 dígitos para municípios (IBGE sem o dígito verificador)
    # Ensure 'Codigo_Municipio' exists before processing
    if 'Codigo_Municipio' in pop_df.columns:
        pop_df['Codigo_Municipio'] = pop_df['Codigo_Municipio'].astype(str)
        # Replace 'None' strings with pandas' NA, then drop rows with NA
        pop_df['Codigo_Municipio'] = pop_df['Codigo_Municipio'].replace('None', pd.NA)
        pop_df = pop_df.dropna(subset=['Codigo_Municipio'])
        # Attempt to convert to int after slicing, handling potential errors
        pop_df['Codigo_Municipio'] = pop_df['Codigo_Municipio'].str[:6].apply(lambda x: int(float(x)) if pd.notna(x) and x.replace('.','').isdigit() else pd.NA)
        pop_df = pop_df.dropna(subset=['Codigo_Municipio'])
        pop_df['Codigo_Municipio'] = pop_df['Codigo_Municipio'].astype(int)
        pop_df = pop_df.drop_duplicates(subset=['Codigo_Municipio'])

    # Converter populacao para número
    if 'Populacao' in pop_df.columns:
        pop_df['Populacao'] = pd.to_numeric(pop_df['Populacao'], errors='coerce')

    # Visualizar o resultado
    print(pop_df.head())

    # 3. Salvar em Excel ou CSV
    # Ensure the directory exists before saving
    output_dir = os.path.dirname("populacao_municipios.csv")
    if output_dir and not os.path.exists(output_dir):
        os.makedirs(output_dir)
    pop_df.to_csv("populacao_municipios.csv", index=False)
else:
    print("Não foi possível carregar os dados de população. DataFrame 'pop_df' está vazio.")
```

```python
import pandas as pd
import json
from datetime import datetime

# Ajuste aqui o nome da sua coluna de semana epidemiológica
col_semana = "SEM_NOT"  # exemplos possíveis: "SEM_NOT", "SEMANA_EP", "SEM_PRI"

ano = 2026
semana_inicial = 1

# Semana epidemiológica mais recente existente na base
semana_final = int(df[col_semana].max())

print(f"Semana epidemiológica considerada: {semana_inicial} a {semana_final}")
```

Agrupar casos por município e semana

```python
# Ensure 'ID_MN_RESI' is numeric and handle missing/invalid values before grouping
df['ID_MN_RESI'] = pd.to_numeric(df['ID_MN_RESI'], errors='coerce')
# Drop rows where 'ID_MN_RESI' is NaN, as these cannot be grouped by municipality ID
df_cleaned = df.dropna(subset=['ID_MN_RESI']).copy()
df_cleaned['ID_MN_RESI'] = df_cleaned['ID_MN_RESI'].astype(int)

casos_semanais = (
    df_cleaned
    .groupby(["ID_MN_RESI", col_semana])
    .size()
    .reset_index(name="casos")
)

display(casos_semanais.head())
```

Gerar documentos temporais por município

```python
documentos_temporais = []

for municipio_id, grupo in casos_semanais.groupby("ID_MN_RESI"):

    grupo = grupo.copy()

    # Criar todas as semanas de 1 até a semana mais recente
    semanas = pd.DataFrame({
        col_semana: range(semana_inicial, semana_final + 1)
    })
    # Convert 'SEM_NOT' in 'semanas' to string to match 'grupo'
    semanas[col_semana] = semanas[col_semana].astype(str)

    # Completar semanas sem registro com zero
    serie = semanas.merge(
        grupo[[col_semana, "casos"]],
        on=col_semana,
        how="left"
    )

    serie["casos"] = serie["casos"].fillna(0).astype(int)

    total_casos = int(serie["casos"].sum())

    # Get the full week code for the peak week
    semana_pico_full_code = serie[col_semana].loc[serie['casos'].idxmax()]
    # Extract last two digits for the week number for metadata and text
    semana_pico = int(str(semana_pico_full_code)[-2:])
    casos_pico = int(serie['casos'].max())

    # Extract the week number from semana_final (which is a full year-week code)
    semana_final_week_number = int(str(semana_final)[-2:])

    # Buscar nome e UF do município, se já estiver em municipios_criticos
    info_municipio = municipios_criticos[
        municipios_criticos["ID_MN_RESI"] == municipio_id
    ]

    if len(info_municipio) > 0:
        nome_municipio = info_municipio.iloc[0]["Nome_Municipio"]
        uf = info_municipio.iloc[0]["UF"]
    else:
        nome_municipio = "Município não identificado"
        uf = "--"

    serie_texto = "; ".join(
        [
            # Extract last two digits for week number in text generation
            f"semana {int(row[col_semana][-2:])}: {int(row['casos'])} casos"
            for _, row in serie.iterrows()
        ]
    )

    doc = {
        "id": f"serie_temporal_municipio_{municipio_id}_{ano}_se{semana_inicial:02d}_se{semana_final_week_number:02d}",
        "metadata": {
            "tipo_evidencia": "SerieTemporalMunicipio",
            "agrupamento": "municipio_semana_epidemiologica",
            "ano": ano,
            "semana_inicial": semana_inicial,
            "semana_final": semana_final_week_number, # Use week number
            "id_municipio": int(municipio_id),
            "municipio": nome_municipio,
            "uf": uf,
            "total_casos_periodo": total_casos,
            "semana_pico": semana_pico, # Use week number
            "casos_semana_pico": casos_pico,
            "fonte": "SINAN",
            "data_geracao": datetime.now().strftime("%Y-%m-%d")
        },
        "text": (
            f"Série temporal de casos prováveis de dengue no município de "
            f"{nome_municipio} ({uf}) durante as semanas epidemiológicas "
            f"{semana_inicial} a {semana_final_week_number} de {ano}. " # Use week number
            f"No período analisado foram registrados {total_casos} casos. "
            f"O maior número semanal ocorreu na semana epidemiológica {semana_pico}, " # Use week number
            f"com {casos_pico} casos. "
            f"A distribuição semanal foi: {serie_texto}."
        )
    }

    documentos_temporais.append(doc)

print(f"Total de documentos temporais gerados: {len(documentos_temporais)}")
```

```python
symptom_cols_decoded = [
    'FEBRE_DECODED',
    'MIALGIA_DECODED',
    'CEFALEIA_DECODED',
    'EXANTEMA_DECODED',
    'VOMITO_DECODED',
    'NAUSEA_DECODED',
    'DOR_COSTAS_DECODED'
]

rag_documents_symptoms_by_state = []

# Group by UF_NAME
grouped_symptoms_by_state = df.groupby('UF_NAME')

for uf_name, group in grouped_symptoms_by_state:
    # Calculate total cases for the current UF
    total_cases = len(group)

    # Initialize symptom counts
    symptom_counts = {}

    # Count occurrences of 'Sim' for each decoded symptom
    for symptom_col in symptom_cols_decoded:
        # Ensure the column exists before trying to count
        if symptom_col in group.columns:
            count_sim = group[symptom_col].eq('Sim').sum()
            # Store the count using a human-readable symptom name
            # Example: 'FEBRE_DECODED' -> 'Febre'
            symptom_name = symptom_col.replace('_DECODED', '').replace('_', ' ').title()
            symptom_counts[symptom_name] = count_sim

    # Construct text_content
    uf_text = uf_name if pd.notna(uf_name) and uf_name != 'Não Informado' else "UF não informada"
    text_content_parts = [f"No estado de {uf_text}, foram notificados {total_cases} casos de dengue."]

    symptom_summary_parts = []
    for symptom_name, count in symptom_counts.items():
        if count > 0:
            symptom_summary_parts.append(f"{count} casos apresentaram {symptom_name.lower()}")

    if symptom_summary_parts:
        text_content_parts.append("Desses casos, " + "; ".join(symptom_summary_parts) + ".")
    else:
        text_content_parts.append("Nenhum sintoma específico foi registrado para esta UF.")

    text_content = " ".join(text_content_parts)

    # Create metadata dictionary
    metadata = {
        "document_type": "symptoms_by_state",
        "uf_name": uf_name if pd.notna(uf_name) else "Não Informado",
        "total_cases": total_cases
    }
    # Add symptom counts to metadata
    metadata.update(symptom_counts)

    # Combine metadata and text_content into a RAG document
    rag_document = {
        "metadata": metadata,
        "text_content": text_content
    }
    rag_documents_symptoms_by_state.append(rag_document)

print(f"Generated {len(rag_documents_symptoms_by_state)} RAG documents for symptom counts by state.")
print("\nExample RAG documents (first 3):\n")
for i, doc in enumerate(rag_documents_symptoms_by_state[:3]):
    print(f"--- Document {i+1} ---")
    print(f"Metadata: {doc['metadata']}")
    print(f"Text Content: {doc['text_content']}")
    print("--------------------\n")
```

```python
sex_mapping = {
    'F': 'Feminino',
    'M': 'Masculino'
}

hospitalization_mapping = {
    '1': 'Sim',
    '2': 'Nao'
}

evolution_mapping = {
    '1': 'Cura',
    '2': 'Obito',
    '9': 'Ignorado'
}

symptom_severity_mapping = {
    '1': 'Sim',
    '2': 'Nao'
}

# Apply mappings to create decoded columns
df['CS_SEXO_DECODED'] = df['CS_SEXO'].map(sex_mapping).fillna('Nao Informado')
df['HOSPITALIZ_DECODED'] = df['HOSPITALIZ'].map(hospitalization_mapping).fillna('Nao Informado')
df['GRAV_CONSC_DECODED'] = df['GRAV_CONSC'].map(symptom_severity_mapping).fillna('Nao Informado')
df['GRAV_ORGAO_DECODED'] = df['GRAV_ORGAO'].map(symptom_severity_mapping).fillna('Nao Informado')
df['EVOLUCAO_DECODED'] = df['EVOLUCAO'].map(evolution_mapping).fillna('Nao Informado')

# Decode common symptoms that are binary (1/2 for Sim/Nao)
symptom_cols = ['FEBRE', 'MIALGIA', 'CEFALEIA', 'EXANTEMA', 'VOMITO', 'NAUSEA', 'DOR_COSTAS', 'CONJUNTVIT', 'ARTRITE', 'ARTRALGIA', 'PETEQUIA_N', 'LEUCOPENIA', 'LACO','DOR_RETRO']
for col in symptom_cols:
    df[f'{col}_DECODED'] = df[col].map(symptom_severity_mapping).fillna('Nao Informado')

print("Decoded columns created. Displaying head of relevant decoded columns:")
display(df[['CS_SEXO_DECODED', 'HOSPITALIZ_DECODED', 'GRAV_CONSC_DECODED', 'GRAV_ORGAO_DECODED', 'EVOLUCAO_DECODED', 'FEBRE_DECODED', 'MIALGIA_DECODED']].head())
```

**Reasoning**:
Now that relevant categorical columns have been decoded, I will define a list of these decoded columns along with age, UF name, and sorotype. Then, I will iterate through each row of the DataFrame to construct a concise textual summary for each dengue case, replacing any 'Nao Informado' or NaN values with more appropriate phrasing like 'não informado' or omitting the detail if it adds no value. Finally, I will display the head of the new 'case_summary' column to verify the output.



```python
relevant_cols_for_summary = [
    'NU_IDADE_N',
    'CS_SEXO_DECODED',
    'UF_NAME',
    'SOROTIPO_DECODED',
    'HOSPITALIZ_DECODED',
    'GRAV_CONSC_DECODED',
    'GRAV_ORGAO_DECODED',
    'COMPLICA_DECODED',
    'EVOLUCAO_DECODED',
    'FEBRE_DECODED',
    'MIALGIA_DECODED',
    'CEFALEIA_DECODED',
    'EXANTEMA_DECODED',
    'VOMITO_DECODED',
    'NAUSEA_DECODED',
    'DOR_COSTAS_DECODED'
]

def generate_case_summary(row):
    summary_parts = []

    # Age and Sex
    age = row['NU_IDADE_N']
    if pd.notna(age):
        summary_parts.append(f"Paciente com {int(age)} anos,")
    else:
        summary_parts.append("Paciente com idade não informada,")

    sex = row['CS_SEXO_DECODED']
    if sex != 'Nao Informado':
        summary_parts.append(f"sexo {sex.lower()}.")
    else:
        summary_parts.append("sexo não informado.")

    # Location and Sorotype
    uf_name = row['UF_NAME']
    sorotipo = row['SOROTIPO_DECODED']
    if pd.notna(uf_name) and pd.notna(sorotipo):
        summary_parts.append(f"Notificado em {uf_name}, sorotipo {sorotipo}.")
    elif pd.notna(uf_name):
        summary_parts.append(f"Notificado em {uf_name}, sorotipo não informado.")
    elif pd.notna(sorotipo):
        summary_parts.append(f"Notificado em UF não informada, sorotipo {sorotipo}.")
    else:
        summary_parts.append("Localização e sorotipo não informados.")

    # Hospitalization
    hospitalized = row['HOSPITALIZ_DECODED']
    if hospitalized == 'Sim':
        summary_parts.append("Houve hospitalização.")
    elif hospitalized == 'Nao':
        summary_parts.append("Não houve hospitalização.")
    # else: "HOSPITALIZ_DECODED" being 'Nao Informado' implies no info, which can be omitted

    # Severity indicators
    grav_consc = row['GRAV_CONSC_DECODED']
    grav_orgao = row['GRAV_ORGAO_DECODED']
    complicacao = row['COMPLICA_DECODED'] # This is likely all NaN from previous step

    severity_info = []
    if grav_consc == 'Sim':
        severity_info.append("alteração da consciência")
    if grav_orgao == 'Sim':
        severity_info.append("atingimento de outros órgãos")
    if complicacao == 'Sim':
        severity_info.append("complicações")

    if severity_info:
        summary_parts.append(f"Sinais de gravidade incluem: {', '.join(severity_info)}.")
    # else: If no severity info, just omit this part

    # Evolution
    evolution = row['EVOLUCAO_DECODED']
    if evolution != 'Nao Informado':
        summary_parts.append(f"Evolução do caso: {evolution.lower()}.")

    # Symptoms
    symptoms = []
    for col in ['FEBRE', 'MIALGIA', 'CEFALEIA', 'EXANTEMA', 'VOMITO', 'NAUSEA', 'DOR_COSTAS']:
        decoded_symptom = row[f'{col}_DECODED']
        if decoded_symptom == 'Sim':
            symptoms.append(col.replace('_', ' ').lower()) # Format symptom name

    if symptoms:
        summary_parts.append(f"Sintomas apresentados: {', '.join(symptoms)}.")
    else:
        summary_parts.append("Sintomas não informados.")

    return ' '.join(summary_parts)

df['case_summary'] = df.apply(generate_case_summary, axis=1)

print("Generated case summaries. Displaying first 5 summaries:")
display(df['case_summary'].head())
```

#Agrupamento por sintomas por UF e ano

```python
symptom_cols_decoded = [
    'FEBRE_DECODED',
    'MIALGIA_DECODED',
    'CEFALEIA_DECODED',
    'EXANTEMA_DECODED',
    'VOMITO_DECODED',
    'NAUSEA_DECODED',
    'DOR_COSTAS_DECODED',
    'CONJUNTVIT_DECODED', 'ARTRITE_DECODED', 'ARTRALGIA_DECODED',
    'PETEQUIA_N_DECODED', 'LEUCOPENIA_DECODED', 'LACO','DOR_RETRO_DECODED'
]

rag_documents_symptoms_by_state = []

# Group by UF_NAME and NU_ANO to include the year
grouped_symptoms_by_state = df.groupby(['UF_NAME', 'NU_ANO'])

for (uf_name, nu_ano), group in grouped_symptoms_by_state:
    # Calculate total cases for the current UF and year
    total_cases = len(group)

    # Initialize symptom counts
    symptom_counts = {}

    # Count occurrences of 'Sim' for each decoded symptom
    for symptom_col in symptom_cols_decoded:
        # Ensure the column exists before trying to count
        if symptom_col in group.columns:
            count_sim = group[symptom_col].eq('Sim').sum()
            # Store the count using a human-readable symptom name, converted to int
            symptom_name = symptom_col.replace('_DECODED', '').replace('_', ' ').title()
            symptom_counts[symptom_name] = int(count_sim)

    # Determine the principal symptom for this group
    principal_symptom = "Não Informado"
    max_symptom_count = -1
    if symptom_counts:
        # Find the symptom with the highest count
        principal_symptom = max(symptom_counts, key=symptom_counts.get)
        max_symptom_count = symptom_counts[principal_symptom]

    # Construct text_content
    uf_text = uf_name if pd.notna(uf_name) and uf_name != 'Não Informado' else "UF não informada"
    ano_text = nu_ano if pd.notna(nu_ano) else "ano não informado"
    text_content_parts = [f"No estado de {uf_text}, no ano de {ano_text}, foram notificados {total_cases} casos de dengue."]

    symptom_summary_parts = []
    for symptom_name, count in symptom_counts.items():
        if count > 0:
            symptom_summary_parts.append(f"{count} casos apresentaram {symptom_name.lower()}")

    if symptom_summary_parts:
        text_content_parts.append("Desses casos, " + "; ".join(symptom_summary_parts) + ".")
    else:
        text_content_parts.append("Nenhum sintoma específico foi registrado para esta UF.")

    if principal_symptom != "Não Informado":
        text_content_parts.append(f"O sintoma principal foi {principal_symptom.lower()} com {max_symptom_count} casos.")

    text_content = " ".join(text_content_parts)

    # Create metadata dictionary, converting numerical values to int
    metadata = {
        "document_type": "symptoms_by_state",
        "uf_name": uf_name if pd.notna(uf_name) else "Não Informado",
        "nu_ano": int(nu_ano) if pd.notna(nu_ano) else "Não Informado",
        "total_cases": int(total_cases),
        "semana_epidemiologica_range": "1 a 11", # Add the epidemiological week range here
        "principal_symptom": principal_symptom, # Add principal symptom to metadata
        "principal_symptom_count": int(max_symptom_count) if max_symptom_count != -1 else "Não Informado"
    }
    # Add symptom counts to metadata
    metadata.update(symptom_counts)

    # Combine metadata and text_content into a RAG document
    rag_document = {
        "metadata": metadata,
        "text_content": text_content
    }
    rag_documents_symptoms_by_state.append(rag_document)

print(f"Generated {len(rag_documents_symptoms_by_state)} RAG documents for symptom counts by state.")
print("\nExample RAG documents (first 3):\n")
for i, doc in enumerate(rag_documents_symptoms_by_state[:3]):
    print(f"--- Document {i+1} ---")
    print(f"Metadata: {doc['metadata']}")
    print(f"Text Content: {doc['text_content']}")
    print("--------------------\n")
```

```python
import google.generativeai as genai
from chromadb.utils import embedding_functions
from openai import OpenAI
import os
import pandas as pd

# Configure o cliente OpenAI para gerar embeddings da query e para a geração da resposta
openai_client = OpenAI(api_key=OPENAI_API_KEY)

def rerank_documents(query, retrieved_documents, openai_client, llm_model="gpt-4o-mini"):
    """
    Reordena os documentos recuperados com base na relevância usando o próprio LLM.

    Args:
        query (str): A pergunta original do usuário.
        retrieved_documents (list): Uma lista de dicionários, onde cada dicionário
                                    representa um documento recuperado e contém 'text_content'
                                    e 'metadata'.
        openai_client: O cliente inicializado da OpenAI.
        llm_model (str): O modelo do LLM a ser usado para reranking.

    Returns:
        list: A lista de documentos reranking ordenados do mais relevante para o menos relevante.
    """

    prompt_rerank_template = """
Você é um especialista em dados de saúde e epidemiologia. Sua tarefa é avaliar a relevância de um trecho de documento para responder a uma pergunta específica.

Pergunta do usuário:
{pergunta}

Trecho do documento:
{texto_documento}

Avalie a relevância desse trecho para responder à pergunta. Responda apenas com um número inteiro de 0 a 10, onde 10 é extremamente relevante e 0 é irrelevante. Não adicione nenhum texto adicional além do número.
"""

    documents_with_scores = []

    for doc_data in retrieved_documents:
        text_content = doc_data['text_content']
        metadata = doc_data['metadata']

        # Construct the rerank prompt for the current document
        rerank_prompt = prompt_rerank_template.format(
            pergunta=query,
            texto_documento=text_content
        )

        try:
            # Call LLM for reranking score
            response = openai_client.chat.completions.create(
                model=llm_model,
                messages=[{"role": "user", "content": rerank_prompt}],
                max_tokens=5,
                temperature=0.0 # Use temperature 0 for consistent scoring
            )
            score_str = response.choices[0].message.content.strip()
            score = float(score_str)
        except Exception as e:
            print(f"Erro ao obter score de reranking para o documento: {e}. Definindo score como 0.")
            score = 0.0

        documents_with_scores.append((score, {'text_content': text_content, 'metadata': metadata}))

    # Sort documents from most relevant to least relevant
    sorted_documents = sorted(documents_with_scores, key=lambda x: x[0], reverse=True)

    return [doc for score, doc in sorted_documents]


def ask_rag_with_chroma(query_text, num_results=5):
    # 1. Criar o embedding da query do usuário
    query_embedding_response = openai_client.embeddings.create(
        input=[query_text],
        model="text-embedding-3-small"
    )
    query_embedding = query_embedding_response.data[0].embedding

    # 2. Recuperar documentos relevantes do ChromaDB (um pouco mais do que o necessário para o reranking)
    initial_num_retrievals = num_results + 3 # Retrieve a few more documents for better reranking candidates
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=initial_num_retrievals,
        include=['documents', 'metadatas'] # Inclua o conteúdo do documento e metadados
    )

    print(f"\n--- Documentos Recuperados Inicialmente para a Query: '{query_text}' ---")
    retrieved_docs_for_rerank = []
    if results and results['documents']:
        for i, doc_list in enumerate(results['documents']):
            for j, doc_content in enumerate(doc_list):
                metadata_item = results['metadatas'][i][j]
                print(f"Document {i*len(doc_list)+j+1} Content: {doc_content[:200]}...") # Print first 200 chars
                print(f"Document {i*len(doc_list)+j+1} Metadata: {metadata_item}")
                retrieved_docs_for_rerank.append({'text_content': doc_content, 'metadata': metadata_item})
    else:
        print("Nenhum documento recuperado inicialmente.")
    print("--------------------------------------------------\n")

    # 3. Aplicar Reranking
    print(f"--- Aplicando Reranking em {len(retrieved_docs_for_rerank)} documentos ---")
    reranked_documents = rerank_documents(query_text, retrieved_docs_for_rerank, openai_client)

    # Use os 'num_results' (ou menos, se houver) documentos reranked para o contexto final
    final_context_docs = reranked_documents[:num_results]

    # Construir o contexto formatado a partir dos documentos reranking
    print(f"--- Documentos Reranked (Top {num_results}) para o Contexto Final ---")
    formatted_context_parts = []
    if final_context_docs:
        for i, doc_data in enumerate(final_context_docs):
            doc_content = doc_data['text_content']
            metadata_item = doc_data['metadata']
            print(f"Reranked Document {i+1} Content: {doc_content[:200]}...")
            print(f"Reranked Document {i+1} Metadata: {metadata_item}")
            formatted_context_parts.append(f"Documento {i+1}:\nConteúdo: {doc_content}\nMetadados: {metadata_item}")
    else:
        print("Nenhum documento reranking para o contexto.")
    print("--------------------------------------------------\n")

    context = "\n\n".join(formatted_context_parts)

    # Construir o prompt para o LLM no formato de mensagens do OpenAI Chat Completion
    messages = [
        {"role": "system", "content": "Você é um assistente útil e informativo. Sua tarefa é responder perguntas sobre dados de dengue com base nas informações fornecidas nos documentos de contexto. Inclua os metadados relevantes para dar credibilidade à resposta. Se a informação não estiver disponível nos documentos, diga que você não tem essa informação."},
        {"role": "user", "content": f"Documentos de contexto:\n{context}\nPergunta: {query_text}"}
    ]

    # 4. Gerar a resposta usando o LLM da OpenAI
    try:
        response = openai_client.chat.completions.create(
            model="gpt-4o-mini",  # Você pode mudar para outro modelo, como "gpt-4o", "gpt-4" etc.
            messages=messages,
            max_tokens=500
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"Ocorreu um erro ao gerar a resposta do LLM: {e}"

# Exemplo de uso:
query = "Quantos casos de dengue foram notificados no Acre em 2026?"
rag_response = ask_rag_with_chroma(query)
print(f"Pergunta: {query}")
print(f"Resposta RAG: {rag_response}")

query_2 = "Quais os sorotipos mais comuns em Minas Gerais no ano de 2026?"
rag_response_2 = ask_rag_with_chroma(query_2)
print(f"\nPergunta: {query_2}")
print(f"Resposta RAG: {rag_response_2}")

query_3 = "Qual o principal sintoma em Pernambuco em 2026?"
rag_response_3 = ask_rag_with_chroma(query_3)
print(f"\nPergunta: {query_3}")
print(f"Resposta RAG: {rag_response_3}")

query_4 = "Qual a semana epidemiológica com mais casos de DEN 2 no Rio de Janeiro?"
rag_response_4 = ask_rag_with_chroma(query_4)
print(f"\nPergunta: {query_4}")
print(f"Resposta RAG: {rag_response_4}")
```

```python
import requests
import numpy as np
import json # Added import for json

def get_municipio_name_from_ibge(municipio_id):
    """
    Fetches the municipality name from the IBGE API given an IBGE municipality ID.
    """
    if pd.isna(municipio_id) or not str(municipio_id).isdigit():
        return "Não Informado"

    # Convert to string and ensure it has 7 digits by padding with zeros if necessary
    # IBGE IDs are 7 digits long, but sometimes they might be shorter in datasets
    municipio_id_str = str(int(municipio_id)).zfill(7)

    # IBGE API endpoint for a specific municipality
    url = f"https://servicodados.ibge.gov.br/api/v1/localidades/municipios/{municipio_id_str}"

    try:
        response = requests.get(url)
        response.raise_for_status()  # Raise an HTTPError for bad responses (4xx or 5xx)
        data = response.json()

        # The API returns a list of dicts if the ID matches multiple items (e.g. for general query)
        # or if it's a specific municipality, it returns a dict directly.
        # We'll assume the first item if it's a list.
        if isinstance(data, list) and data:
            return data[0].get('nome', 'Nome não encontrado')
        elif isinstance(data, dict):
            return data.get('nome', 'Nome não encontrado')
        else:
            return 'Formato de resposta inesperado'

    except requests.exceptions.HTTPError as e:
        # Handle cases where the ID might not exist or other HTTP errors
        if response.status_code == 404:
            return "ID do município não encontrado no IBGE"
        else:
            print(f"HTTP error for ID {municipio_id_str}: {e}")
            return "Erro de API"
    except requests.exceptions.RequestException as e:
        print(f"Request error for ID {municipio_id_str}: {e}")
        return "Erro de conexão"
    except json.JSONDecodeError:
        print(f"JSON decode error for ID {municipio_id_str}. Response: {response.text}")
        return "Erro de processamento de dados"

print("Função 'get_municipio_name_from_ibge' definida.")
```

```python
import pandas as pd

def gerar_relatorio_hospitalizacao_por_municipio(df, caminho_ibge):
    agrupado = (
        df.groupby(['ID_MUNICIP', 'HOSPITALIZ'])
          .size()
          .unstack(fill_value=0)
          .copy()
    )

    agrupado.index = pd.to_numeric(agrupado.index, errors='coerce')
    agrupado = agrupado[agrupado.index.notna()].copy()
    agrupado.index = agrupado.index.astype('int64')
    agrupado = agrupado.reset_index()

    ibge = pd.read_csv(caminho_ibge)
    ibge = ibge.rename(columns={
        'codigo_ibge': 'ID_MUNICIP',
        'nome_municipio': 'Nome_Municipio'
    })

    ibge['ID_MUNICIP'] = pd.to_numeric(ibge['ID_MUNICIP'], errors='coerce')
    ibge = ibge[ibge['ID_MUNICIP'].notna()].copy()
    ibge['ID_MUNICIP'] = ibge['ID_MUNICIP'].astype('int64')
    ibge = ibge.drop_duplicates(subset=['ID_MUNICIP'])

    agrupado = agrupado.merge(
        ibge[['ID_MUNICIP', 'Nome_Municipio']],
        on='ID_MUNICIP',
        how='left'
    )

    agrupado['Nome_Municipio'] = agrupado['Nome_Municipio'].fillna('Não encontrado')

    colunas_esperadas = ['Sim', 'Nao', 'Ignorado', 'Nao_Informado']
    for col in colunas_esperadas:
        if col not in agrupado.columns:
            agrupado[col] = 0

    agrupado['Total_Casos'] = agrupado[colunas_esperadas].sum(axis=1)

    agrupado['Taxa_Internacao_%'] = (
        agrupado['Sim'] / agrupado['Total_Casos'].replace(0, pd.NA) * 100
    ).fillna(0).round(2)

    agrupado = agrupado[
        ['ID_MUNICIP', 'Nome_Municipio', 'Sim', 'Nao', 'Ignorado',
         'Nao_Informado', 'Total_Casos', 'Taxa_Internacao_%']
    ].sort_values(
        by=['Taxa_Internacao_%', 'Total_Casos'],
        ascending=[False, False]
    )

    return agrupado
```

# Gerar o documento em formato JSONL

```python
import json
import os
import pandas as pd

# =========================
# Preparação dos campos temporais
# =========================
df = df.copy()

df["SEM_NOT"] = df["SEM_NOT"].astype(str).str.strip()
df["ano"] = df["SEM_NOT"].str[:4]
df["semana"] = df["SEM_NOT"].str[-2:]

# se preferir usar NU_ANO como fonte principal do ano:
df["ano"] = df["NU_ANO"].astype(str)

# =========================
# Agrupamento por UF + ano
# =========================
agrupado_hospitalizacao_uf_periodo = (
    df.groupby(["SG_UF_NOT", "UF_NAME", "ano"], as_index=False)
      .agg(
          internacoes=("HOSPITALIZ_CAT", lambda x: (x == 'Sim').sum()),
          nao_hospitalizados=("HOSPITALIZ_CAT", lambda x: (x == 'Nao').sum()),
          ignorado=("HOSPITALIZ_CAT", lambda x: (x == 'Ignorado').sum()),
          nao_informado=("HOSPITALIZ_CAT", lambda x: (x == 'Nao_Informado').sum()),
          total_casos=('SG_UF_NOT', 'count'), # Fix: Use (column, 'count') for group size
          semana_min=("semana", "min"),
          semana_max=("semana", "max")
      )
)

# =========================
# Cálculo das taxas
# =========================
agrupado_hospitalizacao_uf_periodo["taxa_internacao_bruta"] = (
    agrupado_hospitalizacao_uf_periodo["internacoes"] /
    agrupado_hospitalizacao_uf_periodo["total_casos"] * 100
).round(2)

validos = (
    agrupado_hospitalizacao_uf_periodo["internacoes"] +
    agrupado_hospitalizacao_uf_periodo["nao_hospitalizados"] +
    agrupado_hospitalizacao_uf_periodo["ignorado"]
)

agrupado_hospitalizacao_uf_periodo["taxa_internacao_valida"] = (
    agrupado_hospitalizacao_uf_periodo["internacoes"] / validos.replace(0, pd.NA) * 100
).round(2).fillna(0)

agrupado_hospitalizacao_uf_periodo["taxa_incompletude"] = (
    agrupado_hospitalizacao_uf_periodo["nao_informado"] /
    agrupado_hospitalizacao_uf_periodo["total_casos"] * 100
).round(2)

# =========================
# Geração dos documentos JSON
# =========================
documentos = []

for _, row in agrupado_hospitalizacao_uf_periodo.iterrows():
    semana_min = str(row["semana_min"]).zfill(2)
    semana_max = str(row["semana_max"]).zfill(2)
    faixa = f"1 a {int(semana_max)}" if semana_min == "01" else f"{int(semana_min)} a {int(semana_max)}"

    metadata = {
        "tipo_documento": "hospitalizacao_por_uf_periodo",
        "uf_sigla": str(row["SG_UF_NOT"]),
        "uf_name": row["UF_NAME"],
        "ano": int(row["ano"]),
        "semana_epidemiologica_range": faixa,
        "semana_final": semana_max,
        "internacoes": int(row["internacoes"]),
        "nao_hospitalizados": int(row["nao_hospitalizados"]),
        "ignorado": int(row["ignorado"]),
        "nao_informado": int(row["nao_informado"]),
        "total_casos": int(row["total_casos"]),
        "taxa_internacao_bruta": float(row["taxa_internacao_bruta"]),
        "taxa_internacao_valida": float(row["taxa_internacao_valida"]),
        "taxa_incompletude": float(row["taxa_incompletude"]),
        "categoria": "hospitalizacao",
        "granularidade": "uf_periodo",
        "uf": str(row["SG_UF_NOT"]),
        "sg_uf_not": str(row["SG_UF_NOT"]),
        "nu_ano": int(row["ano"]),
        "ano_normalizado": int(row["ano"]),
        "semana_normalizada": semana_max,
        "num_cases_normalizado": int(row["total_casos"])
    }

    texto = (
        f"No estado de {row['UF_NAME']} ({row['SG_UF_NOT']}), "
        f"no ano de {int(row['ano'])}, considerando a faixa epidemiológica de {faixa}, "
        f"foram registrados {int(row['internacoes'])} casos hospitalizados, "
        f"{int(row['nao_hospitalizados'])} casos sem hospitalização, "
        f"{int(row['ignorado'])} registros ignorados e "
        f"{int(row['nao_informado'])} não informados. "
        f"O total de casos foi de {int(row['total_casos'])}. "
        f"A taxa bruta de internação foi de {float(row['taxa_internacao_bruta']):.2f}%, "
        f"enquanto a taxa válida foi de {float(row['taxa_internacao_valida']):.2f}%. "
        f"A taxa de incompletude foi de {float(row['taxa_incompletude']):.2f}%."
    )

    documentos.append({
        "metadata": metadata,
        "text_content": texto
    })
```

# Evolução do quadro (óbito) por UF

```python
import pandas as pd

# 1. Agrupar por UF de Notificação e Evolução (Óbitos)
# SG_UF_NOT é o padrão para a sigla do estado no SINAN
# Explicitly convert 'EVOLUCAO' to string to ensure consistent column names after unstack
agrupado_uf = df.groupby(['SG_UF_NOT', df['EVOLUCAO'].astype(str)]).size().unstack(fill_value=0)

# 2. Renomear as colunas para facilitar a visualização
# 2: Óbito pelo agravo, 3: Óbito por outras causas
# Use string keys for the mapping to match the string column names from unstack
colunas_obitos = {
    '1': 'Cura',
    '2': 'Obito_Agravo',
    '3': 'Obito_Outras_Causas',
    '9': 'Ignorado'
}

# The rename operation will only affect columns that exist in the DataFrame.
agrupado_uf = agrupado_uf.rename(columns=colunas_obitos)

# --- Adicionar o nome da UF ---
# Reutilizar o mapeamento de UF já existente ou criar um se não existir
# Para garantir que UF_NAME está disponível, podemos obtê-lo do df original
uf_names_df = df[['SG_UF_NOT', 'UF_NAME']].drop_duplicates().set_index('SG_UF_NOT')
agrupado_uf = agrupado_uf.merge(uf_names_df, left_index=True, right_index=True, how='left')

# Reordenar as colunas para colocar UF_NAME ao lado de SG_UF_NOT (que é o índice agora)
# Primeiro, resetar o índice para que SG_UF_NOT se torne uma coluna
agrupado_uf = agrupado_uf.reset_index()

# Encontrar a posição de 'SG_UF_NOT' e 'UF_NAME'
cols = agrupado_uf.columns.tolist()
sg_uf_not_idx = cols.index('SG_UF_NOT')
uf_name_idx = cols.index('UF_NAME')

# Mover 'UF_NAME' para depois de 'SG_UF_NOT' se não estiver lá
if uf_name_idx != sg_uf_not_idx + 1:
    uf_name_col = cols.pop(uf_name_idx)
    cols.insert(sg_uf_not_idx + 1, uf_name_col)
    agrupado_uf = agrupado_uf[cols]

# 3. Mostrar o resultado ordenado pela quantidade de óbitos pelo agravo
# Add a check to ensure the 'Obito_Agravo' column exists before sorting
if 'Obito_Agravo' in agrupado_uf.columns:
    print(agrupado_uf.sort_values(by='Obito_Agravo', ascending=False))
else:
    print("Coluna 'Obito_Agravo' não encontrada após renomear. Verifique os valores na coluna 'EVOLUCAO'.")
    print("Colunas atuais:", agrupado_uf.columns.tolist())
```

# Salvar o agrupamento no formato JSON para o RAG

```python
import json
import os
import pandas as pd

# Caminho no Google Drive
output_dir = "/content/drive/MyDrive/Doutorado/arbovirus_rag/data_processed/data_rag_gemini"
os.makedirs(output_dir, exist_ok=True)

output_file = os.path.join(output_dir, "docs_obitos_agravo_por_uf_ano_semana.json")

# Filter for 'Obito_Agravo' (EVOLUCAO == '2') and group by UF, Year, and Epidemiological Week
# The original df has NU_ANO (year) and SEM_NOT (epidemiological week)
# EVOLUCAO is type object, so compare with string '2'
obitos_agravos_por_uf_ano_semana_df = \
    df[df['EVOLUCAO'] == '2'].groupby(['NU_ANO', 'SG_UF_NOT', 'UF_NAME', 'SEM_NOT']).size().reset_index(name='Obito_Agravo')

docs_obitos_agravo = []

for _, row in obitos_agravos_por_uf_ano_semana_df.iterrows(): # Use the new DataFrame
    uf = str(row["SG_UF_NOT"]).strip() if "SG_UF_NOT" in row and pd.notna(row["SG_UF_NOT"]) else None
    uf_name = str(row["UF_NAME"]).strip()
    ano = int(row["NU_ANO"]) if pd.notna(row["NU_ANO"]) else None
    # Extract only the last two digits for the week number
    semana_full_code = str(row["SEM_NOT"]).strip() if pd.notna(row["SEM_NOT"]) else None
    semana = semana_full_code[-2:] if semana_full_code else None

    obitos_agravo = int(row["Obito_Agravo"]) if pd.notna(row["Obito_Agravo"]) else 0

    doc = {
        "metadata": {
            "document_type": "obitos_agravo_por_uf",
            "categoria": "mortalidade",
            "granularidade": "uf_semana",
            "uf": uf,
            "uf_name": uf_name,
            "ano": ano,
            "semana_epidemiologica": semana, # This will now be the two-digit week
            "obitos_agravo": obitos_agravo,
            "uf_normalizada": uf if uf else uf_name.lower(),
            "ano_normalizado": ano,
            "semana_normalizada": semana, # This will also be the two-digit week
            "num_obitos_normalizado": obitos_agravo
        },
        "text_content": (
            f"No estado de {uf_name}"
            f"{f' ({uf})' if uf else ''}, "
            f"na semana epidemiológica {semana} do ano de {ano}, "
            f"foram registrados {obitos_agravo} óbitos por agravo de dengue."
        )
    }

    docs_obitos_agravo.append(doc)

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(docs_obitos_agravo, f, ensure_ascii=False, indent=4)

print(f"Arquivo salvo com sucesso em:\n{output_file}")
print(f"Total de documentos gerados: {len(docs_obitos_agravo)}")
```
